```python
validation_report = {
    "rows": len(df1),
    "columns": len(df1.columns),
    "missing_values": df1.isna().sum().sum(),
    "duplicate_rows": df1.duplicated().sum(),
    "duplicate_employee_ids": df1["employee_id"].duplicated().sum(),
    "invalid_ages": ((df1["age"] < 18) | (df1["age"] > 100)).sum(),
    "invalid_salaries": (df1["salary_kes"] <= 0).sum(),
    "invalid_ratings": (
        (df1["performance_rating"] < 1) |
        (df1["performance_rating"] > 5)
    ).sum(),
    "invalid_dates": pd.to_datetime(
        df1["hire_date"], errors="coerce"
    ).isna().sum()
}

validation_report
```

```python
validation_rules = {
    "employee_id": {
        "not_null": True,
        "unique": True,
        "positive": True
    },
    "age": {
        "min": 18,
        "max": 100
    },
    "salary_kes": {
        "positive": True
    },
    "performance_rating": {
        "min": 1,
        "max": 5
    },
    "hire_date": {
        "not_null": True,
        "valid_date": True
    }
}

validation_rules
```

```python
df1["validation_errors"] = ""

# Employee ID checks
df1.loc[df1["employee_id"].isna(), "validation_errors"] += "Missing employee_id; "
df1.loc[df1["employee_id"].duplicated(keep=False), "validation_errors"] += "Duplicate employee_id; "
df1.loc[df1["employee_id"] <= 0, "validation_errors"] += "Invalid employee_id; "

# Age checks
df1.loc[(df1["age"] < 18) | (df1["age"] > 100), "validation_errors"] += "Invalid age; "

# Salary checks
df1.loc[df1["salary_kes"] <= 0, "validation_errors"] += "Invalid salary; "

# Performance rating checks
df1.loc[
    (df1["performance_rating"] < 1) |
    (df1["performance_rating"] > 5),
    "validation_errors"
] += "Invalid performance rating; "

# Hire date checks
df1.loc[df1["hire_date"].isna(), "validation_errors"] += "Missing hire date; "

# Check the results
df1["validation_errors"].value_counts()
```

```python
valid_df = df1[df1["validation_errors"] == ""].copy()

rejected_df = df1[df1["validation_errors"] != ""].copy()

print("Valid records:", len(valid_df))
print("Rejected records:", len(rejected_df))
```

```python
clean_df1 = valid_df.copy()

print("Clean dataset shape:", clean_df1.shape)
print("Rejected records:", len(rejected_df))
```

```python
clean_df1 = clean_df1.drop(columns=["validation_errors"])

print(clean_df1.shape)
```

```python
final_check = {
    "rows": len(clean_df1),
    "columns": len(clean_df1.columns),
    "missing_values": clean_df1.isna().sum().sum(),
    "duplicate_rows": clean_df1.duplicated().sum(),
    "duplicate_employee_ids": clean_df1["employee_id"].duplicated().sum(),
    "invalid_ages": ((clean_df1["age"] < 18) | (clean_df1["age"] > 100)).sum(),
    "invalid_salaries": (clean_df1["salary_kes"] <= 0).sum(),
    "invalid_ratings": (
        (clean_df1["performance_rating"] < 1) |
        (clean_df1["performance_rating"] > 5)
    ).sum(),
    "invalid_dates": pd.to_datetime(
        clean_df1["hire_date"], errors="coerce"
    ).isna().sum()
}

final_check
```
